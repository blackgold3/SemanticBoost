```python
'''
头部向量定义，可以截断连续帧，但必须是单位向量
'''

import os
import numpy as np

root_path = "/data/TTA/motionWE/results/cond_npys"
os.makedirs(root_path, exist_ok=True)
target_name = "head_forward.npy"

condition = np.zeros([100, 3])
condition[:, :] = np.array([1, 0, 1])
scale = 1.0 #### 振幅
condition = condition / (np.linalg.norm(condition, axis=1) + 1e-6)[:, None] ### 方向
condition *= scale
np.save(os.path.join(root_path, target_name), condition)
```

```python
'''
左手腕向量定义，带有振幅和方向，相对于根节点，例如 0.5 * [0, 0, 1] 就是在根节点正前方，0.5 的距离
'''

import os
import numpy as np

root_path = "/data/TTA/motionWE/results/cond_npys"
os.makedirs(root_path, exist_ok=True)
target_name = "leftwrist.npy"

condition = np.zeros([120, 3])
condition[:, :] = np.array([-1, 0, 1])
scale = 0.5 #### 振幅
condition = condition / (np.linalg.norm(condition, axis=1) + 1e-6)[:, None] ### 方向
condition *= scale
np.save(os.path.join(root_path, target_name), condition)
```

```python
'''
右脚腕向量定义，带有振幅和方向，相对于根节点，例如 0.5 * [0, 0, 1] 就是在根节点正前方，0.5 的距离
'''

import os
import numpy as np

root_path = "/data/TTA/motionWE/results/cond_npys"
os.makedirs(root_path, exist_ok=True)
target_name = "rightfoot.npy"

condition = np.zeros([160, 3])
condition[140:160, :] = np.array([0.2, -1.0, 0])
scale = 0.5 #### 振幅
condition = condition / (np.linalg.norm(condition, axis=1) + 1e-6)[:, None] ### 方向
condition *= scale
np.save(os.path.join(root_path, target_name), condition)
```

```python
'''
轨迹向量定义
'''

import os
import numpy as np

root_path = "/data/TTA/motionWE/results/cond_npys"
os.makedirs(root_path, exist_ok=True)
target_name = "trace_left.npy"

# 起始点和结束点
start_point = np.array([0, 0, 0])
end_point = np.array([1.5, 0, 0])
frames = 120
condition = np.linspace(start_point, end_point, frames)
np.save(os.path.join(root_path, target_name), condition)
```
